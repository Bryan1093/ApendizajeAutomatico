# **Tipos Comunes de Gráficos de Distribución:**

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Fetch the dataset
wine_quality = fetch_ucirepo(id=186)

# Data (as pandas DataFrames)
X = wine_quality.data.features
y = wine_quality.data.targets

# Get the frequency of each quality category
frecuencia = y['quality'].value_counts()

# Create the pie chart
plt.figure(figsize=(10, 8))  # Increased figure size
# Explode the slices (e.g., explode the largest slice)
explode = [0.1 if i == frecuencia.idxmax() else 0 for i in frecuencia.index]

# Custom function to display both value and percentage
def make_autopct(values):
    def my_autopct(pct):
        total = sum(values)
        val = int(round(pct*total/100.0))
        return '{v:d} ({p:.1f}%)'.format(v=val, p=pct)
    return my_autopct

# Use a different color palette (e.g., 'viridis') and explode slices
plt.pie(frecuencia, labels=frecuencia.index, autopct=make_autopct(frecuencia), startangle=140,
        colors=plt.cm.viridis(np.linspace(0, 1, len(frecuencia))), explode=explode)

plt.title('Distribución de Categorías de Calidad del Vino')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.legend(frecuencia.index, loc="center left", bbox_to_anchor=(1, 0, 0.5, 1)) # Legend outside

plt.show()
```

**Interpretación del Gráfico de Pastel de la Calidad del Vino:**

El código genera un gráfico de pastel que muestra la distribución de las diferentes categorías de calidad del vino en el dataset "VINOS".

**Elementos del gráfico:**

- **Sectores:** Cada sector del pastel representa una categoría de calidad del vino (3, 4, 5, 6, 7, 8, 9).
- **Tamaño de los sectores:** El tamaño de cada sector es proporcional a la frecuencia (cantidad de vinos) que pertenecen a esa categoría de calidad.
- **Etiquetas:** Cada sector tiene una etiqueta que indica la categoría de calidad que representa.
- **Porcentajes:** Dentro de cada sector, se muestra el porcentaje de vinos que pertenecen a esa categoría, junto con el número absoluto de vinos.
- **Explosión:** El sector que representa la categoría de calidad más frecuente (en este caso, probablemente la calidad 6) estará ligeramente separado del resto del pastel para resaltarlo.
- **Leyenda:** Se mostrará una leyenda que relaciona los colores de los sectores con las categorías de calidad.

**Interpretación general:**

El gráfico de pastel permitirá visualizar rápidamente la proporción de vinos en cada categoría de calidad. Se podrán observar las siguientes tendencias:

- **Categorías predominantes:** Se identificarán fácilmente las categorías de calidad que tienen la mayor cantidad de vinos.
- **Categorías menos frecuentes:** Se podrán observar las categorías de calidad que tienen pocos vinos.
- **Distribución general:** Se podrá obtener una idea general de cómo se distribuye la calidad del vino en el dataset, si está concentrada en ciertas categorías o si es más uniforme.

**Interpretación del Diagrama de Caja de Nivel de Alcohol vs. Calidad del Vino:**

El código genera un diagrama de caja que muestra la relación entre el nivel de alcohol (alcohol) y la calidad (quality) del vino en el dataset "VINOS".

**Elementos del diagrama de caja:**

**Eje X (horizontal):** Representa las diferentes categorías de calidad del vino (3, 4, 5, 6, 7, 8, 9).

**Eje Y (vertical):** Representa el nivel de alcohol en los vinos.

**Cajas:** Cada caja representa la distribución del nivel de alcohol para una categoría de calidad específica.
  - La línea central dentro de la caja representa la mediana.
  - Los bordes superior e inferior de la caja representan el tercer cuartil (Q3) y el primer cuartil (Q1), respectivamente.
  - La altura de la caja (Q3 - Q1) se conoce como rango intercuartílico (IQR).

**Bigotes:** Las líneas que se extienden desde la caja (bigotes) representan la variabilidad de los datos fuera del IQR.
  - Generalmente, los bigotes se extienden hasta 1.5 veces el IQR desde los cuartiles.
**Valores atípicos:** Los puntos individuales fuera de los bigotes se consideran valores atípicos, que son valores inusualmente altos o bajos.

```python
# 3. Scatter Plot of Volatile Acidity vs. Quality:
plt.figure(figsize=(8, 6))
sns.scatterplot(x='volatile_acidity', y='quality', data=df)
plt.title('Acidez Volátil vs. Calidad del Vino')
plt.xlabel('Acidez Volátil')
plt.ylabel('Calidad')
plt.show()
```

**Interpretación del Gráfico de Dispersión de Acidez Volátil vs. Calidad del Vino:**

El gráfico muestra la relación entre la acidez volátil (volatile_acidity) y la calidad (quality) del vino. Cada punto representa un vino.

**Interpretación:**

- **Valores atípicos:** Podrían existir algunos valores atípicos, es decir, vinos con acidez volátil inusualmente alta o baja para su calidad.

El gráfico sugiere que la acidez volátil puede ser un factor importante en la calidad del vino, pero no es el único determinante. Vinos con menor acidez volátil tienden a tener mayor calidad.

**Elementos del gráfico de pares:**

- Matriz de gráficos: El gráfico de pares crea una matriz de gráficos donde:
  - La diagonal principal muestra histogramas de la distribución de cada variable individual.
  - Las celdas fuera de la diagonal principal muestran gráficos de dispersión (scatter plots) de las relaciones entre pares de variables.

- Relaciones entre variables: Los gráficos de dispersión permiten visualizar la relación entre dos variables, mostrando si existe una tendencia lineal, una relación no lineal o si las variables son independientes.
- Distribución de variables: Los histogramas en la diagonal principal muestran la distribución de cada variable individual, permitiendo observar la forma de la distribución, la frecuencia de los valores y la presencia de valores atípicos.

**Interpretación general:**

El gráfico de pares es una herramienta útil para la exploración de datos, ya que permite visualizar rápidamente las relaciones entre múltiples variables. Se pueden observar las siguientes tendencias:

- **Correlaciones:** Se pueden identificar correlaciones entre variables observando la forma de los gráficos de dispersión. Una tendencia lineal ascendente indica una correlación positiva, mientras que una tendencia lineal descendente indica una correlación negativa.
- **Distribuciones:** Los histogramas en la diagonal principal muestran la distribución de cada variable. Se puede observar si la distribución es simétrica, sesgada o si tiene valores atípicos.
- **Relaciones no lineales:** Si los puntos en un gráfico de dispersión no siguen una tendencia lineal, podría indicar una relación no lineal entre las variables.
